```python
import numpy as np 
#alll 0s matrix
np.zeros((2,3))
```

```python
# all 1s is matrix
np.ones((4,2,2), dtype = "int32")
```

```python
#any other number
np.full((2,2), 99)
```

```python
#any other  (fill_line)
a = ([2,3],[3,4])
np.full_like(a,4)
```

```python
#random decimal numbers
np.random.rand(4,2)
#4 rows 2 col
```

```python
#random integer values
np.random.randint(-4,7 ,size=(3,3))
```

```python
np.identity(5)
```

```python
arr = np.array([[1,2,3]])
r1 = np.repeat(arr,3,axis =0)
print(r1)
```

```python
output = np.ones((5,5))
print(output)

z = np.zeros((3,3))
z[1,1] = 9 
print(z)

output[1:-1,1:-1] = z
print(output)
```

```python
## be careful when copying arrays
a = np.array([1,2,3])
b = a #b a=b.copy()
b[0] = 100 #it changes the content also # to prevent use.copy
print(a)
```
